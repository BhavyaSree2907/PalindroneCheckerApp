```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.decomposition import PCA
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_auc_score,
    RocCurveDisplay
)

print("All libraries imported successfully!")
```

```python
df = pd.read_csv("dog_adoption_master.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

df.head()
```

```python
sample_df, _ = train_test_split(
    df,
    train_size=8000,
    random_state=42,
    stratify=df["returned"]
)

X = sample_df.drop(
    columns=[
        "adoption_id",
        "return_reason",
        "days_to_return",
        "returned"
    ]
)
y = sample_df["returned"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

numeric_cols = X_train.select_dtypes(
    include=np.number
).columns.tolist()

categorical_cols = X_train.select_dtypes(
    exclude=np.number
).columns.tolist()

preprocessor = ColumnTransformer([
    ("num", StandardScaler(), numeric_cols),
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols)
])

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print("Preprocessing setup completed!")
```

```python
linear_model = Pipeline([
    ("preprocessing", preprocessor),
    ("pca", PCA(n_components=0.95)),
    ("svm", SVC(
        kernel="linear",
        C=1.0,
        class_weight="balanced"
    ))
])

linear_model.fit(X_train, y_train)

y_pred_linear = linear_model.predict(X_test)

print("Linear SVM trained successfully!")
```

```python
print("LINEAR SVM RESULTS")

print("Accuracy:",
      round(accuracy_score(y_test, y_pred_linear), 4))

print("Precision:",
      round(precision_score(y_test, y_pred_linear,
                            zero_division=0), 4))

print("Recall:",
      round(recall_score(y_test, y_pred_linear,
                         zero_division=0), 4))

print("F1 Score:",
      round(f1_score(y_test, y_pred_linear,
                     zero_division=0), 4))

print("\nClassification Report:")
print(classification_report(
    y_test, y_pred_linear, zero_division=0
))
```

```python
cm = confusion_matrix(y_test, y_pred_linear)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Not returned", "Returned"],
    yticklabels=["Not returned", "Returned"]
)

plt.title("Linear SVM - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

RocCurveDisplay.from_estimator(
    linear_model, X_test, y_test
)
plt.title("Linear SVM - ROC Curve")
plt.show()
```

```python
nonlinear_model = Pipeline([
    ("preprocessing", preprocessor),
    ("pca", PCA(n_components=0.95)),
    ("svm", SVC(
        kernel="rbf",
        C=1.0,
        gamma="scale",
        class_weight="balanced"
    ))
])

nonlinear_model.fit(X_train, y_train)

y_pred_nonlinear = nonlinear_model.predict(X_test)

print("Non-Linear SVM trained successfully!")
```

```python
print("NON-LINEAR SVM RESULTS (RBF)")

print("Accuracy:",
      round(accuracy_score(y_test, y_pred_nonlinear), 4))

print("Precision:",
      round(precision_score(y_test, y_pred_nonlinear,
                            zero_division=0), 4))

print("Recall:",
      round(recall_score(y_test, y_pred_nonlinear,
                         zero_division=0), 4))

print("F1 Score:",
      round(f1_score(y_test, y_pred_nonlinear,
                     zero_division=0), 4))

print("ROC-AUC:",
      round(roc_auc_score(
          y_test,
          nonlinear_model.decision_function(X_test)
      ), 4))

print("\nClassification Report:")
print(classification_report(
    y_test, y_pred_nonlinear, zero_division=0
))
```

```python
cm = confusion_matrix(y_test, y_pred_nonlinear)

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=["Not returned", "Returned"],
    yticklabels=["Not returned", "Returned"]
)

plt.title("Non-Linear SVM (RBF) - Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

RocCurveDisplay.from_estimator(
    nonlinear_model, X_test, y_test
)
plt.title("Non-Linear SVM (RBF) - ROC Curve")
plt.show()
```

```python
comparison = pd.DataFrame({
    "Model": ["Linear SVM", "Non-Linear SVM (RBF)"],
    "Accuracy": [
        accuracy_score(y_test, y_pred_linear),
        accuracy_score(y_test, y_pred_nonlinear)
    ],
    "Precision": [
        precision_score(y_test, y_pred_linear, zero_division=0),
        precision_score(y_test, y_pred_nonlinear, zero_division=0)
    ],
    "Recall": [
        recall_score(y_test, y_pred_linear, zero_division=0),
        recall_score(y_test, y_pred_nonlinear, zero_division=0)
    ],
    "F1 Score": [
        f1_score(y_test, y_pred_linear, zero_division=0),
        f1_score(y_test, y_pred_nonlinear, zero_division=0)
    ]
})

print(comparison.round(4).to_string(index=False))

comparison.set_index("Model").plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Linear vs Non-Linear SVM")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
